```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.datasets import mnist
```

```python
#Load and Prepare Data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values to 0–1
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
# Define the Sequential model
model = Sequential([
    # Flatten: Converts each 28x28 image into a 1D vector of length 784
    Flatten(input_shape=(28, 28)),

    # Dense hidden layer with 128 neurons and ReLU activation
    Dense(128, activation='relu'),

    # Output layer with 10 neurons (one for each digit 0–9) and softmax activation
    Dense(10, activation='softmax')
])
```

```python
#Compile the Model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
#Train the Model
model.fit(x_train, y_train, epochs=5)
```

```python
#Evaluate on Unseen Data
test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy:", test_acc)
```

```python
#Predict Individual Samples
predictions = model.predict(x_test)
print("Model predicted:", predictions[0].argmax())
print("Actual value:", y_test[0])
```
