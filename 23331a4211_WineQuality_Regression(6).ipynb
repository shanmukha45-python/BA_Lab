```python
 import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn import metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
red=pd.read_csv('winequality-red.csv',sep=';')
white=pd.read_csv('winequality-white.csv',sep=';')
red['type'] = 0
white['type'] = 1
# Merge datasets
df = pd.concat([red, white], axis=0).reset_index(drop=True)

print('Dataset shape:', df.shape)
print(df.head())
```

```python
x=df.drop('quality',axis=1)
y=df['quality']
```

```python
from sklearn.ensemble import RandomForestRegressor
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
rf_model=RandomForestRegressor()
rf_model.fit(x_train,y_train)
```

```python
#Predict Quality using new values
#new_data = [[7.4,0.70,0.00,1.9,0.076,11.0,34.0,0.99780,3.51,0.56,9.4,0]]
#predicted_quality = rf_model.predict(new_data)
#print("Predicted Quality:", predicted_quality)
```

```python
y_pred_rf = rf_model.predict(x_test)
mae_rf = metrics.mean_absolute_error(y_test, y_pred_rf)
print("Random Forest MAE:", mae_rf)
mse_rf = metrics.mean_squared_error(y_test, y_pred_rf)
print("Random Forest MSE:", mse_rf)
r2_rf = metrics.r2_score(y_test, y_pred_rf)
print("Random Forest R2:", r2_rf)
print("Random Forest RMSE:", np.sqrt(mse_rf))
```
